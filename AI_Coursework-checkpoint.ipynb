```python
import pandas as pd
df = pd.read_csv("diabetes_prediction_dataset-checkpoint.csv")
print("Dataset loaded successfully!")
print("Shape (rows, columns):", df.shape)

df.head()
```

```python
print("Columns before dropping:")
print(df.columns)

# Drop unnamed index column if it exists
if "Unnamed: 0" in df.columns:
    df = df.drop(columns=["Unnamed: 0"])
    print("Dropped column: Unnamed: 0")
else:
    print("No unnecessary columns found.")

print("\nColumns after dropping:")
print(df.columns)
```

```python
df["gender"] = df["gender"].map({"Female": 0, "Male": 1})

print("Gender conversion done.")
print(df["gender"].value_counts(dropna=False))
```

```python
df["smoking_history"] = df["smoking_history"].apply(
    lambda x: 0 if x in ["never", "No Info"] else 1
)

print("Smoking history conversion done.")
print(df["smoking_history"].value_counts(dropna=False))
```

```python
missing_values = df.isnull().sum()
print("Missing values per column:\n")
print(missing_values)
```

```python
num_cols = ["age", "bmi", "HbA1c_level", "blood_glucose_level",
            "hypertension", "heart_disease"]

for col in num_cols:
    df[col] = df[col].fillna(df[col].mean())

# Categorical columns (after conversion)
df["gender"] = df["gender"].fillna(df["gender"].mode()[0])
df["smoking_history"] = df["smoking_history"].fillna(df["smoking_history"].mode()[0])

print("\nMissing values after imputation:\n")
print(df.isnull().sum())
```

```python
### Missing values handles as the data set was checked for missing values
# Missing values in numerical features were filled using the mean and categorical features using the mode.
```

```python
#  Check unique values in categorical columns
print("Unique values in gender:", df["gender"].unique())
print("Unique values in smoking_history:", df["smoking_history"].unique())

# Logical consistency checks for numerical columns# 3) Target consistency
print("Age < 0:", (df["age"] < 0).sum())
print("BMI < 0:", (df["bmi"] < 0).sum())
print("HbA1c < 0:", (df["HbA1c_level"] < 0).sum())
print("Blood glucose < 0:", (df["blood_glucose_level"] < 0).sum())

# Target consistency
print("Invalid diabetes values:",
      (~df["diabetes"].isin([0, 1])).sum())
```

```python

numeric_cols = ["age", "bmi", "HbA1c_level", "blood_glucose_level"]

def count_iqr_outliers(s):
    q1 = s.quantile(0.25)
    q3 = s.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    return ((s < lower) | (s > upper)).sum()

for col in numeric_cols:
    print(f"{col} outliers:", count_iqr_outliers(df[col]))
```

```python
X = df[FEATURES]
y = df[TARGET]
```

```python
y.value_counts()
```

```python

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,       
    stratify=y,          
    random_state=42
)
```

```python
X_train_enc = pd.get_dummies(X_train, columns=CATEGORICAL_FEATURES, drop_first=True)
X_test_enc  = pd.get_dummies(X_test,  columns=CATEGORICAL_FEATURES, drop_first=True)

# Align columns so train and test have the same set of features
X_train_enc, X_test_enc = X_train_enc.align(X_test_enc, join="left", axis=1, fill_value=0)
```

```python
from sklearn.linear_model import LogisticRegression

log_reg = LogisticRegression(max_iter=1000)
log_reg.fit(X_train_enc, y_train)

y_pred_lr = log_reg.predict(X_test_enc)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Logistic Regression Accuracy:", accuracy_score(y_test, y_pred_lr))
print("\nClassification Report:\n", classification_report(y_test, y_pred_lr))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred_lr))
```

```python
from sklearn.tree import DecisionTreeClassifier

# Create model
dt_model = DecisionTreeClassifier(random_state=42)

# Train model
dt_model.fit(X_train_enc, y_train)
```

```python
y_pred_dt = dt_model.predict(X_test_enc)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

print("Decision Tree Accuracy:", accuracy_score(y_test, y_pred_dt))
print("\nClassification Report:\n", classification_report(y_test, y_pred_dt))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred_dt))
```

```python
from sklearn.ensemble import RandomForestClassifier

# Create model (100 trees is standard and matches your proposal idea)
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train model
rf_model.fit(X_train_enc, y_train)
```

```python
y_pred_rf = rf_model.predict(X_test_enc)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

print("Random Forest Accuracy:", accuracy_score(y_test, y_pred_rf))
print("\nClassification Report:\n", classification_report(y_test, y_pred_rf))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred_rf))
```

```python
# Model comparison summary

results = {
    "Logistic Regression": 0.9625,
    "Decision Tree": 0.9625,
    "Random Forest": 0.9775
}

for model, acc in results.items():
    print(f"{model}: {acc*100:.2f}%")
```

```python
from sklearn.metrics import recall_score

# Compute recall for positive class (diabetes = 1)
recall_lr = recall_score(y_test, y_pred_lr, pos_label=1)
recall_dt = recall_score(y_test, y_pred_dt, pos_label=1)
recall_rf = recall_score(y_test, y_pred_rf, pos_label=1)

print("Recall (Diabetes=1) Comparison:")
print(f"Logistic Regression Recall: {recall_lr:.3f}")
print(f"Decision Tree Recall: {recall_dt:.3f}")
print(f"Random Forest Recall: {recall_rf:.3f}")
```
